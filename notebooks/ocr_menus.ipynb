```python
import ollama

# Chemin vers ton image

prompt = """Je souhaite que tu me dresses la liste du menu au format strict JSON du format suivant : {"name":"nom_du_plat","price":prix_en_euros}. Le menu est en chinois traditionnel avec les prix en TWD. Voici l'image du menu :"""
# Envoie la requête à Ollama
response = ollama.generate(
    model="qwen3-vl:235b-cloud",  # Remplace par le nom exact de ton modèle
    prompt=prompt,
    images=[image_path]  # Liste des chemins d'images
)

# Affiche la réponse
print("Réponse du modèle :")
print(response["response"])
```
